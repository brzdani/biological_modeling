# Density-Dependent Natural Selection

**Mathematics of Sections 1 & 2** — after J. Roughgarden, *Ecology* **52**(3): 453–468 (1971)

## 1. General one-locus discrete-generation model

### 1.1 Setup and symbols

Consider a randomly mating population with discrete, non-overlapping generations. Adults are censused at the moment of interbreeding. A single autosomal locus with two alleles **A** and **B** determines the trait of interest. The three genotypes are $AA$, $AB$, and $BB$.

| Symbol | Meaning |
|---|---|
| $N^i$ | Total number of breeding adults in generation $i$ |
| $N^i_{AA},\,N^i_{AB},\,N^i_{BB}$ | Number of adults of each genotype |
| $p^i,\,q^i$ | Frequency of allele A and B, with $p^i+q^i=1$ |
| $D^i,\,H^i,\,R^i$ | Genotypic frequencies $N^i_{AA}/N^i$, $N^i_{AB}/N^i$, $N^i_{BB}/N^i$ |
| $2m$ | Average number of zygotes per mating (genotype-independent) |
| $\ell_{AA},\,\ell_{AB},\,\ell_{BB}$ | Average survival fraction of each genotype from zygote to breeding adult |

Allele frequencies are

$$p^i = D^i + \tfrac{1}{2}H^i, \qquad q^i = R^i + \tfrac{1}{2}H^i. \tag{1}$$

### 1.2 Selective values

Because fertility $m$ is assumed genotype-independent, the product of fertility and survival appears together for each genotype. Roughgarden defines the **selective value** of a genotype as that product:

$$W_{AA}(N,p) = \ell_{AA}(N,p)\,m(N,p),\quad W_{AB}(N,p) = \ell_{AB}(N,p)\,m(N,p),\quad W_{BB}(N,p) = \ell_{BB}(N,p)\,m(N,p). \tag{2}$$

These are *absolute* selective values in the sense of Sewall Wright (1959). They are empirically measured, and depend on population density $N$ and allele frequency $p$. This is the source of both *density* dependence and *frequency* dependence.

### 1.3 Derivation of the recursion

Under random mating, the probability that a randomly chosen mating is $AA\times AA$ is $(D^i)^2$, that it is $AA\times AB$ is $2D^iH^i$, and so on. Since each of the $N^i$ individuals mates on average once, there are $N^i/2$ matings. The Mendelian ratios then give the number of zygotes of each genotype. For example, the number of $AA$ zygotes is

$$N^i m \left[(D^i)^2 + D^i H^i + \tfrac{1}{4}(H^i)^2\right] = N^i m \left[D^i + \tfrac{1}{2}H^i\right]^2 = N^i m\,(p^i)^2.$$

Multiplying by the survival fraction $\ell_{AA}$ and using the definition of $W_{AA}$ in (2) yields the number of $AA$ breeding adults in generation $i+1$:

$$N_{AA}^{i+1} = W_{AA}(N^i,p^i)\,(p^i)^2\,N^i,$$

$$N_{AB}^{i+1} = W_{AB}(N^i,p^i)\,2p^i q^i\,N^i,$$

$$N_{BB}^{i+1} = W_{BB}(N^i,p^i)\,(q^i)^2\,N^i. \tag{3}$$

### 1.4 Change in allele frequency and population size

By the definition of $p$ in (1), the frequency of A in generation $i+1$ is

$$p^{i+1} = \frac{N^{i+1}_{AA} + \tfrac{1}{2}N^{i+1}_{AB}}{N^{i+1}} = \frac{p^i W_{AA}(N^i,p^i) + q^i W_{AB}(N^i,p^i)}{\overline{W}(N^i,p^i)}\,p^i,$$

where the **mean selective value** is

$$\overline{W}(N^i,p^i) = (p^i)^2 W_{AA} + 2p^i q^i W_{AB} + (q^i)^2 W_{BB}. \tag{4}$$

Summing the three genotypes in (3) gives the total population size:

$$N^{i+1} = \overline{W}(N^i,p^i)\,N^i.$$

Rewriting both recursions in *difference* form yields the coupled dynamical system:

$$\Delta p^i = p^{i+1}-p^i = \frac{p^i W_{AA}+q^i W_{AB}-\overline{W}}{\overline{W}}\,p^i, \tag{5}$$

$$\Delta N^i = N^{i+1}-N^i = \bigl[\overline{W}(N^i,p^i)-1\bigr]\,N^i. \tag{6}$$

**Remark.** Equations (5) and (6) are the central result of Section 1. A single set of parameters — the three selective value functions $W_{AA}(N,p)$, $W_{AB}(N,p)$, $W_{BB}(N,p)$ — predicts both gene-frequency evolution and population growth simultaneously.

## 2. Linear density-dependent selective values

### 2.1 The model assumption

The simplest model of density dependence is to let each selective value decrease *linearly* with population size. Write each line in the form $W(N) = a - bN$, but re-parametrize using $(r,K)$:

$$W_{AA}(N) = (r_A + 1) - \frac{r_A}{K_A}N,$$

$$W_{AB}(N) = (r_H + 1) - \frac{r_H}{K_H}N,$$

$$W_{BB}(N) = (r_B + 1) - \frac{r_B}{K_B}N, \tag{7}$$

with the ordering

$$r_A > r_B, \qquad K_B > K_A.$$

Thus allele **A** is the *high-$r$* allele and **B** is the *high-$K$* allele. At low densities the high-$r$ phenotype is fitter; at high densities the high-$K$ phenotype is fitter. The two lines cross at a density $N_e$ defined by $W_{AA}(N_e) = W_{BB}(N_e)$.

### 2.2 Interpretation of $r$ and $K$

Suppose the locus is fixed for allele A, i.e. $p=1$. Substituting (7) into the population-size equation (6) gives

$$N^{i+1} = W_{AA}(N^i)\,N^i = \Bigl[(r_A+1) - \tfrac{r_A}{K_A}N^i\Bigr]N^i,$$

or, in difference form,

$$\Delta N = r_A N\,\frac{K_A - N}{K_A}. \tag{8}$$

This is exactly the **logistic difference equation**, with $r_A$ as the intrinsic rate of increase and $K_A$ as the carrying capacity. The same argument gives $(r_B, K_B)$ when the locus is fixed for B. For stability of the equilibrium $N = K$ in the discrete logistic map, we require

$$0 < r < 1.$$

So each selective value has a *dual interpretation*: as a measure of phenotype fitness (evolutionary), and as a determinant of population growth (demographic).

### 2.3 Geometric picture

In the $(N, W)$ plane, the two homozygote selective values are straight lines with intercepts $(r_A+1)$ and $(r_B+1)$ at $N=0$, and with $x$-intercepts at $K_A$ and $K_B$. The line for A starts higher (higher $r_A$) but hits the axis first (smaller $K_A$); the line for B starts lower but extends further. Their intersection $N_e$ is the density at which the two homozygotes have equal selective value.

## 3. Logistic interpretation of $r$ and $K$

Combining the general recursion (6) with the linear selective values (7) gives, for the homozygote cases:

| Locus fixed for | Selective value | Difference equation |
|---|---|---|
| AA | $W_{AA}(N)$ | $\Delta N = r_A N (K_A - N)/K_A$ |
| BB | $W_{BB}(N)$ | $\Delta N = r_B N (K_B - N)/K_B$ |

For the heterozygote AB, the same logic gives a logistic-like growth with parameters $(r_H, K_H)$, though these may not correspond to a true-breeding strain in practice.

**Consistency check.** Setting $\Delta N = 0$ in (8) gives the non-trivial equilibrium $N = K$. The requirement $0 < r < 1$ ensures that this equilibrium is asymptotically stable in the discrete-time logistic map.

## 4. Heterozygote fitness cases

The evolutionary outcome depends on where the heterozygote line $W_{AB}(N)$ lies relative to the two homozygote lines. Roughgarden classifies 13 cases. The most important ones are summarized below.

| Case | Heterozygote property | Typical outcome (constant environment) |
|---|---|---|
| 1 | Same as high-$r$ homozygote | Fixation of high-$K$ allele |
| 2 | Same as high-$K$ homozygote | Fixation of high-$K$ allele |
| 3–5 | Superior $K$ | Stable polymorphism |
| 6–8 | Inferior $K$ | Bistability: fixation of either allele, depending on initial $p$ |
| 9–10 | Intermediate $K$ | Similar to full-dominance cases |
| 11 | Intermediate at all densities | Fixation of high-$K$ allele |
| 12 | Superior at some densities | Fixation, but with trajectories bent toward center |
| 13 | Inferior at some densities | Fixation, but with trajectories bent away from center |

Formally, the location of $W_{AB}(N)$ relative to the intersection point $(N_e, W_e)$ of $W_{AA}$ and $W_{BB}$ determines the case. If the line passes above the intersection, there is a range of $N$ where the heterozygote is superior; if below, there is a range where it is inferior; if through the point, the heterozygote is intermediate at every $N$ (except $N_e$).

## 5. Coupled dynamical system used in simulation

The full model used in the simulations is the pair of difference equations

$$\begin{aligned}
p^{i+1} &= \frac{p^i W_{AA}(N^i) + q^i W_{AB}(N^i)}{\overline{W}(N^i)}\,p^i, \\[4pt]
N^{i+1} &= \overline{W}(N^i)\,N^i, \\[4pt]
\overline{W}(N^i) &= (p^i)^2 W_{AA}(N^i) + 2p^i q^i W_{AB}(N^i) + (q^i)^2 W_{BB}(N^i),
\end{aligned} \tag{9}$$

with the linear selective values (7). This is the system integrated in the companion Python script.

The phase plane is especially informative: trajectories typically rise through the region $N < N_e$ (where the high-$r$ allele is favored) and then move horizontally at roughly constant $N$ toward the high-$K$ corner as $N > K_A$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def W_AA(N):
    return (r_A + 1.0) - (r_A / K_A) * N

def W_AB(N):
    return (r_H + 1.0) - (r_H / K_H) * N

def W_BB(N):
    return (r_B + 1.0) - (r_B / K_B) * N
```

```python
# ------------------------------------------------------------
# Section 1: General discrete-generation equations
# p' = p * (p W_AA + q W_AB) / W_bar
# N' = W_bar * N
# W_bar = p^2 W_AA + 2 p q W_AB + q^2 W_BB
# ------------------------------------------------------------
def simulate(N0, p0, generations):
    N = np.zeros(generations + 1)
    p = np.zeros(generations + 1)
    N[0] = N0
    p[0] = p0

    for i in range(generations):
        wAA = W_AA(N[i])
        wAB = W_AB(N[i])
        wBB = W_BB(N[i])
        q = 1.0 - p[i]

        w_bar = (p[i]**2) * wAA + 2.0 * p[i] * q * wAB + (q**2) * wBB

        # allele frequency update (Eq. 6)
        p[i+1] = p[i] * (p[i] * wAA + q * wAB) / w_bar

        # population size update (Eq. 8)
        N[i+1] = w_bar * N[i]

    return N, p
```

```python
# ------------------------------------------------------------
# Simulation parameters
# ------------------------------------------------------------
N0 = 100.0          # initial population size
p0 = 0.1            # initial frequency of allele A
generations = 100
```

```python
# ------------------------------------------------------------
# Section 2: Linear density-dependent selective values
# W_ij(N) = (r_ij + 1) - (r_ij / K_ij) * N
# ------------------------------------------------------------

r_A, K_A = 0.8, 8000      # high-r homozygote AA
r_B, K_B = 0.6, 12000     # high-K homozygote BB
r_H, K_H = 0.7, 10000     # heterozygote AB (intermediate case)

N, p = simulate(N0, p0, generations)

# ------------------------------------------------------------
# Plotting: three panels
# ------------------------------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

# Panel 1: allele frequency vs time
axes[0].plot(range(generations + 1), p, 'b-', linewidth=1.8)
axes[0].set_xlabel('Generation')
axes[0].set_ylabel('Allele frequency  p')
axes[0].set_title('Allele frequency over time')
axes[0].grid(True, alpha=0.3)

# Panel 2: population size vs time
axes[1].plot(range(generations + 1), N, 'r-', linewidth=1.8)
axes[1].set_xlabel('Generation')
axes[1].set_ylabel('Population size  N')
axes[1].set_title('Population size over time')
axes[1].grid(True, alpha=0.3)

# Panel 3: phase plane N vs p
axes[2].plot(p, N, 'k.-', markersize=3, linewidth=1.0)
axes[2].plot(p[0], N[0], 'go', markersize=8, label='start')
axes[2].plot(p[-1], N[-1], 'rs', markersize=8, label='end')
axes[2].set_xlabel('Allele frequency  p')
axes[2].set_ylabel('Population size  N')
axes[2].set_title('Phase plane: N vs p')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
